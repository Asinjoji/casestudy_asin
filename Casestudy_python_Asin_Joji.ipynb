Problem: Employee Performance Analyzer

```python
employees = [
    {"id":101, "name":"Alice", "department":"Data", "salary":65000, "rating":4.5},
    {"id":102, "name":"Bob", "department":"AI", "salary":72000, "rating":4.8},
    {"id":103, "name":"Charlie", "department":"Web", "salary":58000, "rating":3.9},
    {"id":104, "name":"David", "department":"Data", "salary":61000, "rating":4.2},
    {"id":105, "name":"Eva", "department":"AI", "salary":75000, "rating":4.9},
    {"id":106, "name":"Frank", "department":"Cloud", "salary":68000, "rating":4.1},
    {"id":107, "name":"Grace", "department":"Web", "salary":59000, "rating":3.8},
    {"id":108, "name":"Helen", "department":"Data", "salary":70000, "rating":4.6}
]
print(f"Total employees = {len(employees)}\n")
print(f"Average salary = {sum(employee['salary'] for employee in employees) / len(employees)}")
print(f"Highest salary = {max(employee['salary'] for employee in employees)}")
print(f"Lowest salary = {min(employee['salary'] for employee in employees)}\n")
print(f"Employees with Salary > 65000 and Rating >= 4.5: ")
for employee in employees:
    if employee['salary'] > 65000 and employee['rating'] > 4.5:
        print(employee['name'])
print("\n")

employee_dict = {101:"Alice",102:"Bob",103:"Charlie",104:"David",105:"Eva",106:"Frank",107:"Grace",108:"Helen"}
print(f"Employee Dictionary \n{employee_dict}")
print("\n")

names = ["Alice","Bob","Charlie","David","Eva","Frank","Grace","Helen"]
print(f"Employee List \n{names}")
names.sort()
print(f"Alphabetically: \n{names}")
names.reverse()
print(f"Reverse:\n{names}")
print(f"First three employee names:")
for i in range (3):
  print(names[i])
print("\n")

employee_id=(101,102,103,104,105,106,107,108)
print(f"Employee ID tuple")
print(employee_id)
print("\n")
print(f"Maximum Employee ID = {max(employee_id)}")
print(f"Minimum Employee ID = {min(employee_id)}")
print("\n")

department={"Data","AI","Web","Cloud"}
print(f"departments \n{department}")
print("\n")

employee_id=int(input("Enter Employee ID: "))
for employee in employees:
  if employee['id'] == employee_id:
    print("Employee Found")
    print(f"ID: {employee['id']}")
    print(f"Name: {employee['name']}")
    print(f"Department: {employee['department']}")
    print(f"Salary: {employee['salary']}")
    print(f"Rating: {employee['rating']}")
    break
else:
  print("Employee not found")
```
